Experiment 10: A Decision Tree is an algorithm that mimics human decision-making (like a flowchart of "If-Else"
rules). Your task is to implement a simplified Decision Tree Classifier entirely from scratch.
You will use the famous Kaggle Iris flower dataset. The model will look at flower petal and sepal
measurements to classify the exact species of the flower. To build the tree, you will program a
mathematical concept called "Gini Impurity," which calculates how mixed up the data is so the
algorithm knows the best place to split it. Finally, you will print out the structure of the tree so
you can visually see the rules it created, and discuss which feature was the most important.
Note for Students: Building a full infinite-depth tree from scratch is thousands of lines of code.
We will build a clean, maximum depth of 2 (a root and its branches) to understand the core
logic!

```python
import pandas as pd
import numpy as np

# ==========================================
# MAIN EXECUTION (Decision Tree From Scratch)
# ==========================================

print("--- Experiment 10: Decision Tree Classifier ---\n")

# Step 1: Load Iris Data
url = "https://gist.githubusercontent.com/netj/8836201/raw/6f9306ad21398ea43cba4f7d537619d0e07d5ae3/iris.csv"

df = pd.read_csv(url)

# Convert species names to numbers (0, 1, 2)
df['variety'] = df['variety'].astype('category').cat.codes

data = df.values


# ==========================================
# Helper Function: Calculate Gini Impurity
# ==========================================

def calculate_gini(groups, classes):

    total_instances = sum([len(group) for group in groups])
    gini = 0.0

    for group in groups:

        size = float(len(group))

        if size == 0:
            continue

        score = 0.0

        # Check the proportion of each class in this group
        for class_val in classes:
            p = [row[-1] for row in group].count(class_val) / size
            score += p * p

        # Weight the group score by its size
        gini += (1.0 - score) * (size / total_instances)

    return gini


# ==========================================
# Helper Function: Split Data
# ==========================================

def test_split(index, value, dataset):

    left, right = list(), list()

    for row in dataset:

        if row[index] < value:
            left.append(row)
        else:
            right.append(row)

    return left, right


# ==========================================
# Step 2: Find the Best Split
# ==========================================

def get_best_split(dataset):

    class_values = list(set(row[-1] for row in dataset))

    best_index = 999
    best_value = 999
    best_score = 999
    best_groups = None

    # Loop through every feature and every value
    for index in range(len(dataset[0]) - 1):

        for row in dataset:

            groups = test_split(index, row[index], dataset)

            gini = calculate_gini(groups, class_values)

            # If this split has the lowest impurity, save it
            if gini < best_score:

                best_index = index
                best_value = row[index]
                best_score = gini
                best_groups = groups

    return {
        'index': best_index,
        'value': best_value,
        'groups': best_groups
    }


# ==========================================
# Step 3: Create a Single Split / Root Node
# ==========================================

features = list(df.columns[:-1])

root_node = get_best_split(data)

split_feature = features[root_node['index']]
split_value = root_node['value']


# ==========================================
# Step 4: Display Tree and Feature Importance
# ==========================================

print("--- Tree Visualization ---")

print(f"[Root Node] If {split_feature} is < {split_value}:")

print(f" --> Branch Left: Contains {len(root_node['groups'][0])} flowers")

print("[Root Node] Else:")

print(f" --> Branch Right: Contains {len(root_node['groups'][1])} flowers")


print("\n--- Feature Importance Discussion ---")

print(
    f"The algorithm tested all 4 features, but chose "
    f"'{split_feature}' as the Root Node."
)

print(
    "In Decision Trees, the feature used at the very top "
    "(Root) is considered the most important feature."
)

print(
    "This is because it provides the highest Information Gain "
    "or the best separation between the species."
)
```
